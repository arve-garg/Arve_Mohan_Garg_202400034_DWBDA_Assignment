```python
file_path = "/content/GroceryDataset.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 10 Rows:")
display(df.head(10))
```

```python
print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())
```

```python
transactions = []

for i in range(len(df)):

    transaction = []

    for item in df.iloc[i]:

        if pd.notna(item):

            item = str(item).strip()

            if item != "":
                transaction.append(item)

    transactions.append(transaction)

print("Transactions created successfully!")

print("\nFirst 10 Transactions:")

for transaction in transactions[:10]:
    print(transaction)
```

```python
transactions = [
    transaction
    for transaction in transactions
    if len(transaction) > 0
]

print("Total Transactions:")
print(len(transactions))
```

```python
te = TransactionEncoder()

encoded_array = te.fit(
    transactions
).transform(transactions)

basket = pd.DataFrame(
    encoded_array,
    columns=te.columns_
)

print("Transaction Encoding Completed!")

print("\nEncoded Basket:")
display(basket.head(10))

print("\nBasket Shape:")
print(basket.shape)
```

```python
item_frequency = basket.sum()

item_frequency = item_frequency.sort_values(
    ascending=False
)

print("Top 20 Most Frequently Purchased Items:")

display(item_frequency.head(20))
```

```python
top_items = item_frequency.head(10)

plt.figure(figsize=(10, 6))

top_items.plot(
    kind="bar"
)

plt.title("Top 10 Most Frequently Purchased Products")

plt.xlabel("Products")

plt.ylabel("Number of Transactions")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()

plt.show()
```

```python
frequent_itemsets = apriori(
    basket,
    min_support=0.01,
    use_colnames=True
)

frequent_itemsets["length"] = (
    frequent_itemsets["itemsets"].apply(len)
)

frequent_itemsets = frequent_itemsets.sort_values(
    by="support",
    ascending=False
)

print("Frequent Itemsets:")

display(
    frequent_itemsets.head(20)
)
```

```python
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.1
)

print("Total Association Rules:")
print(len(rules))

print("\nAssociation Rules:")

display(rules.head(20))
```

```python
important_rules = rules[
    [
        "antecedents",
        "consequents",
        "support",
        "confidence",
        "lift"
    ]
].copy()

print("Important Association Rule Measures:")

display(important_rules.head(20))
```

```python
important_rules["antecedents"] = (
    important_rules["antecedents"]
    .apply(lambda x: ", ".join(sorted(list(x))))
)

important_rules["consequents"] = (
    important_rules["consequents"]
    .apply(lambda x: ", ".join(sorted(list(x))))
)

important_rules = important_rules.sort_values(
    by="lift",
    ascending=False
)

print("Readable Association Rules:")

display(
    important_rules.head(20)
)
```

```python
strong_rules = important_rules[
    (important_rules["lift"] > 1) &
    (important_rules["confidence"] >= 0.1)
].copy()

strong_rules = strong_rules.sort_values(
    by=["lift", "confidence"],
    ascending=False
)

print("Strong Association Rules:")

display(
    strong_rules.head(20)
)
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    important_rules["support"],
    important_rules["confidence"],
    s=important_rules["lift"] * 30,
    alpha=0.6
)

plt.xlabel("Support")

plt.ylabel("Confidence")

plt.title(
    "Market Basket Analysis: Support vs Confidence"
)

plt.grid(True)

plt.tight_layout()

plt.show()
```

```python
print("=" * 60)
print("        MARKET BASKET ANALYSIS RESULTS")
print("=" * 60)

print("\nTotal Transactions:")
print(basket.shape[0])

print("\nTotal Unique Products:")
print(basket.shape[1])

print("\nTotal Frequent Itemsets:")
print(len(frequent_itemsets))

print("\nTotal Association Rules:")
print(len(important_rules))

print("\nTotal Strong Association Rules:")
print(len(strong_rules))

print("\nTop 10 Strongest Association Rules:")

display(
    strong_rules[
        [
            "antecedents",
            "consequents",
            "support",
            "confidence",
            "lift"
        ]
    ].head(10)
)
```

```python
if len(strong_rules) > 0:

    best_rule = strong_rules.iloc[0]

    print("Strongest Association Rule:")

    print(
        best_rule["antecedents"],
        "→",
        best_rule["consequents"]
    )

    print(
        "Support:",
        round(best_rule["support"], 4)
    )

    print(
        "Confidence:",
        round(best_rule["confidence"], 4)
    )

    print(
        "Lift:",
        round(best_rule["lift"], 4)
    )

else:

    print("No strong association rules found.")
```

```python
strong_rules.to_csv(
    "/content/MBA_Association_Rules.csv",
    index=False
)

frequent_itemsets.to_csv(
    "/content/MBA_Frequent_Itemsets.csv",
    index=False
)

print("Files saved successfully!")

print("/content/MBA_Association_Rules.csv")
print("/content/MBA_Frequent_Itemsets.csv")
```
